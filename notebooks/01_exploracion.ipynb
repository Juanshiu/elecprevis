Corre primero `python src/download_data.py` y `python src/preprocessing.py` desde la raíz del repo.

```python
import sys
sys.path.append("..")

import pandas as pd
import matplotlib.pyplot as plt

from src.preprocessing import run_pipeline

df = run_pipeline(save=False)
df.head()
```

```python
# Patrón horario promedio de consumo
hourly = df.groupby(df.index.hour)["Global_active_power"].mean()
hourly.plot(figsize=(10, 4), title="Consumo promedio por hora del día")
plt.xlabel("Hora")
plt.ylabel("kW")
plt.show()
```

```python
# Patrón semanal
weekly = df.groupby(df.index.dayofweek)["Global_active_power"].mean()
weekly.plot(kind="bar", figsize=(8, 4), title="Consumo promedio por día de la semana (0=lunes)")
plt.ylabel("kW")
plt.show()
```
